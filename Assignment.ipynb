## While loop & Control Statements :
(else, break, continue, pass)

```python
random.randint(1,10)

secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
  guess = int(input("Guess the number (between 1 and 10): "))

  if guess < 1 or guess > 10:
    print( "ohh sorry buddy out of range. Please guess the  number between 1 and 10.")
    continue

  if guess == secret_number:
    print("wow! thats nice You guessed the correct number.")
    break
  elif guess < secret_number:
    print("Too low. Try again.")
  else:
    print("Too high. Try again.")

  attempts -= 1
else:
  print(" today's quota over,Better luck next time!")
```

## Function: BMI Calculator

```python
def function_BMI(height,weight):
    BMI = (weight / (height/100)**2)
    return BMI
height = float(input("enter your current height in cm:"))
weight = int(input("enter your current weight in kg:"))
BMI = function_BMI(height,weight)
print(f"your BMI is {BMI:.2f}")
```
